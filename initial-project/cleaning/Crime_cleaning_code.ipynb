```python
import numpy as np
import pandas as pd

NYPD_data = pd.read_csv("FULL_NYPD_Complaint_Data_Historic_20251112.csv")

NYPD_data.head()

NYPD_data.shape
```

```python
NYPD_data['CMPLNT_FR_DT'] = pd.to_datetime(
    NYPD_data['CMPLNT_FR_DT'],
    errors='coerce',
    format='%m/%d/%Y'
)

oldest_date = NYPD_data['CMPLNT_FR_DT'].min()
oldest_date

felony_df = NYPD_data[NYPD_data["LAW_CAT_CD"] == "FELONY"]

felony_df["OFNS_DESC"].unique()

list(felony_df["OFNS_DESC"].unique())
```

```python
violent_offenses = [
    'RAPE',
    'ROBBERY',
    'FELONY ASSAULT',
    'BURGLARY',
    'MURDER & NON-NEGL. MANSLAUGHTER',
    'SEX CRIMES',
    'DANGEROUS WEAPONS',
    'ARSON',
    'KIDNAPPING & RELATED OFFENSES',
    'HOMICIDE-NEGLIGENT,UNCLASSIFIE',
    'HOMICIDE-NEGLIGENT-VEHICLE',
    'FELONY SEX CRIMES',
    'KIDNAPPING',
    'KIDNAPPING AND RELATED OFFENSES'
]

violent_felony_df = felony_df[felony_df["OFNS_DESC"].isin(violent_offenses)]

violent_felony_df.head()
```

```python
violent_felony_df.loc[:, "CMPLNT_FR_DT"] = pd.to_datetime(
    violent_felony_df["CMPLNT_FR_DT"],
    errors="coerce",
    format="%m/%d/%Y"
)


violent_felony_df = violent_felony_df[
    (violent_felony_df["CMPLNT_FR_DT"] >= "2000-01-01") &
    (violent_felony_df["CMPLNT_FR_DT"] <= "2019-12-31")
]

violent_felony_df["CMPLNT_FR_DT"].min(), violent_felony_df["CMPLNT_FR_DT"].max()
```

```python
violent_felony_df["CMPLNT_FR_DT"] = pd.to_datetime(
    violent_felony_df["CMPLNT_FR_DT"], errors="coerce"
)

cutoff = pd.Timestamp("2013-08-01")

violent_felony_df["time_period"] = np.where(
    violent_felony_df["CMPLNT_FR_DT"] <= cutoff,
    "before Patternizr",
    "after Patternizr"
)

violent_felony_df["time_period"].value_counts()
```

```python
violent_felony_df = violent_felony_df.dropna()
violent_felony_df.isna().sum()
violent_felony_df["time_period"].value_counts()
violent_felony_df.head()
```

```python
violent_felony_df.to_csv("violent_felony_cleaned.csv", index=False)
```
